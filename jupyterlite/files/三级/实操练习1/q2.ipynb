```
import random
```

```
def simulate_test():
    total_videos = 100
    detected_violations = 0
    false_positives = 0
    false_negatives = 0
    results = []

    for i in range(total_videos):
        # 模拟安全帽佩戴的真实情况：假设70%的图像安全帽正确佩戴
        has_violation = random.random() < 0.7

        # 模拟检测情况
        if has_violation:
            # 检测正确的概率为80%
            if random.random() < 0.8:
                detected_violations += 1
                results.append((i+1, "正确检测"))
            else:
                false_negatives += 1
                results.append((i+1, "漏检"))
        else:
            # 对于安全帽违规情况，误报概率为10%
            if random.random() < 0.1:
                false_positives += 1
                results.append((i+1, "误报"))
            else:
                results.append((i+1, "正常"))

    # 计算性能指标
    precision = detected_violations / (detected_violations + false_positives) if (detected_violations + false_positives) else 0
    recall = detected_violations / (detected_violations + false_negatives) if (detected_violations + false_negatives) else 0

    # 输出测试报告
    print("基于计算机视觉智能安全帽佩戴检查系统")
    print("==============================")
    print(f"测试数据总数: {total_videos}")
    print(f"安全帽正确佩戴的数据: {int(total_videos*0.7)}")
    print(f"正确检测安全帽佩戴: {detected_violations}")
    print(f"误报数: {false_positives}")
    print(f"漏检数: {false_negatives}")
    print(f"准确率 (Precision): {precision:.2f}")
    print(f"召回率 (Recall): {recall:.2f}")
```

```
if __name__ == "__main__":
    simulate_test()
```

```
# === 显式调用（确保 Run All 能看到输出）===
simulate_test()
```
