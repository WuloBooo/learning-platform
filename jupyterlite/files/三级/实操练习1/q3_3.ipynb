```
import torch
import torch.nn as nn
```

```
class DummyTrashNet(nn.Module):
    def __init__(self):
        super(DummyTrashNet, self).__init__()
        self.fc = nn.Linear(28*28*3, 5)
    def forward(self, x):
        x = x.view(x.size(0), -1)
        return self.fc(x)
```

```
model = DummyTrashNet()
```

```
def simulate_trash_classification():
    dummy_input = torch.randn(1, 3, 28, 28)
    output = model(dummy_input)
    predicted = torch.argmax(output, dim=1).item()
    return f"模拟垃圾分类结果：{predicted}"
```

```
if __name__ == "__main__":
    result = simulate_trash_classification()
    print(result)
```
