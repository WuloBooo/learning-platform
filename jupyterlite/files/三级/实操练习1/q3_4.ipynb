```
knowledge_base = {
    "垃圾识别": "指利用人工智能技术自动识别垃圾的类别，并指导正确分类投放的系统或工具。",
    "模型优化": "可采用 Adam 优化器与学习率调度器实现模型优化。",
    "交互方式": "通过虚拟数据展示预测结果，可用于验证系统效果。",
    "知识库构建": "知识库可以存储常见问题和解决方案，便于快速查询。"
}
```

```
def query_knowledge(keyword):
    results = []
    for key, value in knowledge_base.items():
        if keyword.lower() in key.lower() or keyword.lower() in value.lower():
            results.append(f"{key}: {value}")
    if not results:
        return "未找到相关知识，请尝试其他关键词。"
    return "\n".join(results)
```

```
if __name__ == "__main__":
    keyword = "垃圾分类"  # 已由转换脚本替换 input()，避免阻塞；可自行修改
    print("查询结果：")
    print(query_knowledge(keyword))
```
